EXPERIMENT 1

Context-Free Grammar Rule Validation and Parse Tree Construction

Aim:
To design a Python program to validate sentences using Context-Free Grammar
(CFG) production rules and construct parse tree representations for valid
sentences.

Grammar:

S → NP VP
NP → Det N
VP → V NP
Det → the | a
N → dog | cat | bone
V → chased | found

```python
# Experiment 1: CFG Rule Validation and Parse Tree Construction

# Define the terminal symbols according to the given CFG
grammar = {
    "Det": ["the", "a"],
    "N": ["dog", "cat", "bone"],
    "V": ["chased", "found"]
}


def parse_cfg(sentence):
    """
    Validate the sentence using the given CFG
    and construct a parse tree for valid sentences.
    """

    # Convert the sentence into lowercase words
    words = sentence.lower().split()

    # According to the grammar:
    # S -> NP VP
    # NP -> Det N
    # VP -> V NP
    #
    # Therefore, a valid sentence has the structure:
    # Det N V Det N

    # Check whether the sentence contains exactly 5 words
    if len(words) != 5:
        return "Invalid Sentence"

    det1, noun1, verb, det2, noun2 = words

    # Check the first determiner
    if det1 not in grammar["Det"]:
        return "Invalid Sentence"

    # Check the first noun
    if noun1 not in grammar["N"]:
        return "Invalid Sentence"

    # Check the verb
    if verb not in grammar["V"]:
        return "Invalid Sentence"

    # Check the second determiner
    if det2 not in grammar["Det"]:
        return "Invalid Sentence"

    # Check the second noun
    if noun2 not in grammar["N"]:
        return "Invalid Sentence"

    # Construct the parse tree
    parse_tree = f"""
S
├── NP
│   ├── Det → {det1}
│   └── N → {noun1}
└── VP
    ├── V → {verb}
    └── NP
        ├── Det → {det2}
        └── N → {noun2}
"""

    return parse_tree
```

```python
# Test the CFG parser using the given test cases

test_cases = [
    "the dog chased a cat",
    "a cat found the bone",
    "dog chased cat",
    "the bone found a dog",
    "chased the dog bone"
]

# Run the parser for each test case
for sentence in test_cases:
    print("Input:", sentence)

    result = parse_cfg(sentence)

    if result == "Invalid Sentence":
        print("Output:", result)
    else:
        print("Output: Valid Parse Tree")
        print(result)

    print("-" * 50)
```

EXPERIMENT 2

Agreement and Feature Structure Checking

Aim:
To design a Python program that verifies subject–verb agreement using
feature structures containing Number and Person attributes.

Feature Rules:

he → singular, third person
she → singular, third person
it → singular, third person
we → plural

eats → singular verb
sleeps → singular verb
eat → plural verb
sleep → plural verb

```python
# Experiment 2: Agreement and Feature Structure Checking

# Define feature structures for subjects.
# Each subject has Number and Person attributes.
subject_features = {
    "he": {
        "number": "singular",
        "person": "third"
    },
    "she": {
        "number": "singular",
        "person": "third"
    },
    "it": {
        "number": "singular",
        "person": "third"
    },
    "we": {
        "number": "plural",
        "person": "first"
    }
}

# Define feature structures for verbs.
# Each verb has a Number attribute.
verb_features = {
    "eats": {
        "number": "singular"
    },
    "sleeps": {
        "number": "singular"
    },
    "eat": {
        "number": "plural"
    },
    "sleep": {
        "number": "plural"
    }
}


def check_agreement(subject, verb):
    """
    Check whether the subject and verb agree
    according to their Number features.
    """

    # Convert the input to lowercase
    subject = subject.lower()
    verb = verb.lower()

    # Check whether the subject exists
    if subject not in subject_features:
        return False

    # Check whether the verb exists
    if verb not in verb_features:
        return False

    # Get the Number feature of the subject
    subject_number = subject_features[subject]["number"]

    # Get the Number feature of the verb
    verb_number = verb_features[verb]["number"]

    # Agreement is true when both Number features match
    return subject_number == verb_number
```

```python
# Test the subject–verb agreement function
# using the test cases given in the assessment.

test_cases = [
    ("he", "eats"),
    ("he", "eat"),
    ("we", "eat"),
    ("we", "sleeps"),
    ("she", "sleeps")
]

# Execute each test case
for subject, verb in test_cases:

    result = check_agreement(subject, verb)

    print("Subject:", subject)
    print("Verb:", verb)
    print("Agreement:", result)
    print("-" * 40)
```

EXPERIMENT 3

Probabilistic Context-Free Grammar (PCFG) Parsing

Aim:
To implement a Python program to parse sentences and compute sentence
probabilities using probabilistic grammar production rules.

PCFG Grammar:

S → NP VP [1.0]
NP → Tom [0.7]
NP → Ann [0.3]
VP → sings [0.4]
VP → dances [0.6]

Probability Formula:

P(Sentence) =
P(S → NP VP) × P(NP) × P(VP)

```python
# Experiment 3: Probabilistic Context-Free Grammar (PCFG) Parsing

# Define the probabilities of the PCFG production rules
grammar_probabilities = {
    "S": {
        "NP VP": 1.0
    },
    "NP": {
        "Tom": 0.7,
        "Ann": 0.3
    },
    "VP": {
        "sings": 0.4,
        "dances": 0.6
    }
}


def sentence_probability(sentence):
    """
    Parse a two-word sentence using the given PCFG
    and calculate its probability.
    """

    # Split the sentence into words
    words = sentence.split()

    # The given grammar generates a two-word sentence
    if len(words) != 2:
        return 0.0

    subject = words[0]
    verb = words[1]

    # Check whether the subject can be generated by NP
    if subject not in grammar_probabilities["NP"]:
        return 0.0

    # Check whether the verb can be generated by VP
    if verb not in grammar_probabilities["VP"]:
        return 0.0

    # Get the probability of S -> NP VP
    s_probability = grammar_probabilities["S"]["NP VP"]

    # Get the probability of the subject
    np_probability = grammar_probabilities["NP"][subject]

    # Get the probability of the verb
    vp_probability = grammar_probabilities["VP"][verb]

    # Calculate sentence probability:
    # P(Sentence) = P(S -> NP VP) × P(NP) × P(VP)
    probability = (
        s_probability *
        np_probability *
        vp_probability
    )

    return probability
```

```python
# Test the PCFG sentence probability function
# using the test cases given in the assessment.

test_cases = [
    "Tom sings",
    "Tom dances",
    "Ann sings",
    "Ann dances",
    "Sam sings"
]

# Execute each test case
for sentence in test_cases:

    probability = sentence_probability(sentence)

    print("Input:", sentence)
    print("Probability:", f"{probability:.2f}")
    print("-" * 40)
```
